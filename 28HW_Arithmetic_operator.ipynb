### Different types of operations performing using Arithmetic Operator on data types

#### Fundamental Data Types 
##### 1)- int : Used to store whole numbers without decimal points.
##### 2)- float : Used to store numbers with decimal points
##### 3)- str : Used to store text or a sequence of characters.
##### 4)- complex : Used to store complex numbers containing real and imaginary parts
##### 5)- bool : Used to represent logical values, either "true" or "false".

#### Arithmetic Operators
##### --- Operator used to perform mathematical calculation on values.
##### 1)-'+' Addition : Adds two values. 
##### 2)-'-' Subtraction : Subtracts one value from another
##### 3)-'*' Multiplication : Multiplies two values.
##### 4)-'/' Division : Divides one value by another value and returns its QUOTIENT.
##### 5)-'//' Floor Division : Divides two values and returns its QUOTIENT but only int value.
##### 6)-'%' Modulus / Modulo : Returns the remainder after division.


#### 1) Integer with different data types
##### - a) int with int 
##### -Two integer values are used to perform different arithmetic operations


```python
a=10
b=5
print("Add",a+b)
print("Sub",a-b)
print("Mult",a*b)
print("Div",a/b)
print("Floor Div",a//b)
print("Mod",a%b)
```

##### b) int with float -
##### - Arithmetic operations between int and float, result is generally a float-point values.

```python
a=10
b=2.5
print("Add",a+b)
print("Sub",a-b)
print("Mult",a*b)
print("Div",a/b)
print("Floor Div",a//b)
print("Mod",a%b)
```
